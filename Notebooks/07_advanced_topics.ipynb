## Decorators
Decorators are a powerful feature in Python that allow you to modify or extend the behavior of functions or methods without changing their source code. They provide a clean and concise way to add functionality such as logging, caching, authentication, or performance monitoring to existing code. In this advanced tutorial, we'll explore various advanced techniques for using decorators in Python, including creating decorators with arguments, stacking decorators, class decorators, and best practices.

### Creating Decorators with Arguments:
Decorators can accept arguments, enabling them to be customized for different use cases. To create a decorator with arguments, you need to define a function that returns a decorator function.


```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(3)
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
```

### Stacking Decorators:
You can apply multiple decorators to a single function, stacking them one on top of the other. When stacking decorators, the order in which they are applied matters, with the innermost decorator being applied first.


```python
def uppercase(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()
    return wrapper

def exclamation(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return f"{result}!"
    return wrapper

@exclamation
@uppercase
def greet(name):
    return f"Hello, {name}"

print(greet("Alice"))
```

### Class Decorators:
Decorators can also be implemented as classes rather than functions. To create a class decorator, you need to implement the __call__() method, which allows instances of the class to be called as functions.


```python
class Timer:
    def __init__(self, func):
        self.func = func

    def __call__(self, *args, **kwargs):
        import time
        start_time = time.time()
        result = self.func(*args, **kwargs)
        end_time = time.time()
        print(f"Execution time: {end_time - start_time} seconds")
        return result

@Timer
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(35))
```

### Chaining Decorators with functools.wraps:
When creating decorators, it's important to preserve metadata such as function name and docstring. You can achieve this by using the functools.wraps decorator from the standard library.


```python
from functools import wraps

def uppercase(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()
    return wrapper

@uppercase
def greet(name):
    """Greet someone."""
    return f"Hello, {name}"

print(greet.__name__)      # Output: greet
print(greet.__doc__)       # Output: Greet someone.
```

### Best Practices:
-   Use decorators to separate concerns and keep code clean and modular.
-   Document your decorators and preserve metadata using functools.wraps.
-   Be cautious when stacking decorators, as the order of application can affect the behavior of the decorated function.


## Generators
Generators are a powerful feature in Python for creating iterators. They allow you to generate a sequence of values lazily, one at a time, instead of generating them all at once and storing them in memory. Generators are efficient for handling large datasets and can be used to create infinite sequences. In this advanced tutorial, we'll explore various advanced techniques for using generators in Python, including generator expressions, generator delegation, coroutine generators, and best practices.

### Generator Expressions:
Generator expressions are a concise way to create generators using a syntax similar to list comprehensions. They are especially useful when you need to iterate over large datasets without loading them all into memory.


```python
# Generator expression to generate squares of numbers from 1 to 10
squares = (x**2 for x in range(1, 11))

for x in squares:
    print(x)
```

### Generator Delegation:
Generator delegation allows one generator to delegate part of its operations to another generator using the yield from syntax. This enables you to compose generators and separate concerns.


```python
def numbers():
    yield 1
    yield 2
    yield 3

def squares():
    yield from (x**2 for x in numbers())

for square in squares():
    print(square)
```

### Coroutine Generators:
Generators can be used as coroutines, which are functions that can yield values in addition to receiving values. This allows for bidirectional communication between coroutines and enables advanced asynchronous programming patterns.


```python
def echo():
    while True:
        received = yield
        print("Received:", received)

coro = echo()
next(coro)
coro.send("Hello")
```

### Asynchronous Generators (Python 3.6+):
Asynchronous generators are a combination of generators and asynchronous programming. They allow you to generate asynchronous sequences of values and are useful for asynchronous data processing.


```python
async def async_generator():
    for i in range(10):
        await asyncio.sleep(1)
        yield i

async for value in async_generator():
    print(value)
```

### Best Practices:
-   Use generator expressions for lazy evaluation of sequences, especially with large datasets.
-   Leverage generator delegation to compose generators and separate concerns, improving code modularity.
-   Explore coroutine generators and asynchronous generators for advanced asynchronous programming tasks.


## Context managers
Context managers are a powerful tool in Python for managing resources and ensuring cleanup actions are performed reliably, especially in scenarios where resources need to be acquired and released safely. Context managers are commonly used with the with statement and can be implemented using classes or generator functions. In this advanced tutorial, we'll explore various advanced techniques for using context managers in Python, including custom context managers, contextlib module, asynchronous context managers, and best practices.

### Custom Context Managers Using Classes:
Context managers can be created by implementing classes with __enter__() and __exit__() methods. The __enter__() method is called when entering the with block, and the __exit__() method is called when exiting the with block.


```python
class Timer:
    def __enter__(self):
        self.start_time = time.time()
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        end_time = time.time()
        print(f"Execution time: {end_time - self.start_time} seconds")

with Timer():
    time.sleep(1)
```

### Custom Context Managers Using Generator Functions:
Context managers can also be implemented using generator functions and the contextlib module. The contextlib.contextmanager decorator simplifies the creation of context managers from generator functions.


```python
from contextlib import contextmanager

@contextmanager
def timer():
    start_time = time.time()
    yield
    end_time = time.time()
    print(f"Execution time: {end_time - start_time} seconds")

with timer():
    time.sleep(1)
```

### Nested Context Managers:
You can nest multiple context managers within a single with statement to manage multiple resources simultaneously. Nested context managers are executed in the order they are specified, with the innermost context manager's __exit__() method called first.


```python
with open("file1.txt") as file1, open("file2.txt") as file2:
    # Read from file1 and file2 simultaneously
    pass
```

### Asynchronous Context Managers (Python 3.7+):
Asynchronous context managers are used for managing asynchronous resources and are created similarly to synchronous context managers, but with asynchronous __aenter__() and __aexit__() methods.


```python
class AsyncTimer:
    async def __aenter__(self):
        self.start_time = time.time()

    async def __aexit__(self, exc_type, exc_value, traceback):
        end_time = time.time()
        print(f"Execution time: {end_time - self.start_time} seconds")

async with AsyncTimer():
    await asyncio.sleep(1)
```

### Context Managers as Decorators:
Context managers can be used as decorators using the contextlib.contextmanager decorator. This allows you to create lightweight context managers without defining a separate class.


```python
from contextlib import contextmanager

@contextmanager
def timer():
    start_time = time.time()
    yield
    end_time = time.time()
    print(f"Execution time: {end_time - start_time} seconds")

@timer
def foo():
    time.sleep(1)

foo()
```

### Best Practices:
-   Use context managers to manage resources and ensure cleanup actions are performed reliably.
-   Prefer using with statements over manual resource management to improve code readability and maintainability.
-   Leverage contextlib module for creating lightweight context managers using generator functions.
-   Consider asynchronous context managers for managing asynchronous resources in asynchronous code.


## Regular expressions
Regular expressions are a powerful tool for pattern matching and text manipulation in Python. They provide a concise and flexible syntax for searching, matching, and extracting text based on specified patterns. In this advanced tutorial, we'll explore various advanced techniques for using regular expressions in Python, including advanced pattern matching, group capturing, lookahead and lookbehind assertions, performance optimizations, and best practices.

### Advanced Pattern Matching:
Regular expressions support a wide range of pattern matching capabilities beyond simple literals. Advanced patterns include:
-   Character classes ([...]): Match any character within the specified set.
-   Quantifiers (*, +, ?, {m,n}): Specify the number of repetitions of a pattern.
-   Anchors (^, $): Match the beginning or end of a line.
-   Alternation (|): Match either of multiple patterns.
-   Escaping special characters: Escape special characters like . or * with \ to match them literally.


```python
import re

# Match email addresses
pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
```

### Group Capturing:
Regular expressions allow you to define groups within patterns using parentheses (). Capturing groups can be used to extract specific parts of a matched string.


```python
# Extract username and domain from email addresses
match = re.match(r'(\b[A-Za-z0-9._%+-]+)@([A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b)', email)
username, domain = match.groups()
```

### Lookahead and Lookbehind Assertions:
Lookahead and lookbehind assertions allow you to specify conditions for matching without including the matched text in the result. Positive lookahead ((?=...)) and negative lookahead ((?!...)) assert conditions ahead of the current position, while positive lookbehind ((?<=...)) and negative lookbehind ((?<!...)) assert conditions behind the current position.


```python
# Match only if followed by a specific pattern
pattern = r'\b\w+(?=ing)\b'
```

### Performance Optimization:
Regular expressions can be computationally expensive, especially for complex patterns or large input strings. You can improve performance by:
-   Precompiling regular expressions using re.compile() for reuse.
-   Using more specific patterns to reduce backtracking.
-   Avoiding unnecessary use of quantifiers like .* when more specific patterns can be used.


```python
# Precompile regular expression for reuse
pattern = re.compile(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b')
```

### Best Practices:
-   Use raw strings (r'...') for regular expression patterns to avoid unintended escape sequences.
-   Precompile regular expressions for performance-critical applications or when using the same pattern multiple times.
-   Test regular expressions thoroughly with various input strings to ensure correct behavior and performance.
