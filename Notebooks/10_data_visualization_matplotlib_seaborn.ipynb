# Data visualization with Matplotlib and Seaborn
Data visualization is a crucial aspect of data analysis, enabling you to gain insights by representing data graphically. This tutorial will focus on data manipulation using Pandas and data visualization using Matplotlib and Seaborn.
## Introduction to Pandas, Matplotlib, and Seaborn
-   Pandas: A powerful data manipulation and analysis library.
-   Matplotlib: A comprehensive library for creating static, animated, and interactive visualizations in Python.
-   Seaborn: A statistical data visualization library based on Matplotlib, providing a high-level interface for drawing attractive and informative statistical graphics.

### Data Preparation with Pandas
We'll start by preparing some sample data using Pandas.


```python
import pandas as pd

# Sample DataFrame for demonstration
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Salary': [50000, 60000, 55000, 65000, 62000],
    'Department': ['HR', 'Engineering', 'HR', 'Engineering', 'Marketing']
}

df = pd.DataFrame(data)
print(df)
```

### Basic Plotting with Matplotlib
Matplotlib is the foundation for data visualization in Python.

#### Line Plot:


```python
import matplotlib.pyplot as plt

# Line plot of Age vs Salary
plt.figure(figsize=(8, 6))
plt.plot(df['Age'], df['Salary'], marker='o')
plt.title('Age vs Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.grid(True)
plt.show()
```

#### Bar Plot:

```python
# Bar plot of Salary by Name
plt.figure(figsize=(8, 6))
plt.bar(df['Name'], df['Salary'], color='skyblue')
plt.title('Salary by Name')
plt.xlabel('Name')
plt.ylabel('Salary')
plt.show()
```

#### Histogram:

```python
# Histogram of Age
plt.figure(figsize=(8, 6))
plt.hist(df['Age'], bins=5, color='green', edgecolor='black')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

## Advanced Plotting with Seaborn
Seaborn provides a higher-level interface for creating visually appealing and informative statistical graphics.

#### Scatter Plot:


```python
import seaborn as sns

# Scatter plot of Age vs Salary
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Age', y='Salary', data=df, hue='Department', style='Department', s=100)
plt.title('Age vs Salary by Department')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```

#### Box Plot:

```python
# Box plot of Salary by Department
plt.figure(figsize=(8, 6))
sns.boxplot(x='Department', y='Salary', data=df)
plt.title('Salary Distribution by Department')
plt.xlabel('Department')
plt.ylabel('Salary')
plt.show()
```

#### Count Plot:

```python
# Count plot of Employees in each Department
plt.figure(figsize=(8, 6))
sns.countplot(x='Department', data=df, palette='viridis')
plt.title('Number of Employees by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

## Combining Pandas with Matplotlib and Seaborn for Data Visualization
Pandas integrates well with both Matplotlib and Seaborn, allowing you to visualize data directly from DataFrames.

### Plotting with Pandas Built-in Plotting Methods:


```python
# Line plot using Pandas built-in plotting
df.plot(kind='line', x='Age', y='Salary', marker='o', figsize=(8, 6), title='Age vs Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.grid(True)
plt.show()

# Bar plot using Pandas built-in plotting
df.plot(kind='bar', x='Name', y='Salary', color='skyblue', figsize=(8, 6), title='Salary by Name')
plt.xlabel('Name')
plt.ylabel('Salary')
plt.show()
```

### Using Seaborn with Pandas:

```python
# Pair plot of the DataFrame
sns.pairplot(df, hue='Department', markers=['o', 's', 'D'])
plt.show()

# Heatmap of correlation matrix
plt.figure(figsize=(8, 6))
corr_matrix = df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

### Basic Plot Types
Let's start with some basic plot types that are frequently used in data visualization.

#### Line Plot
Line plots are useful for visualizing data over a continuous interval or time period.

```python
import plotly.graph_objs as go

# Sample data
x = [0, 1, 2, 3, 4, 5]
y = [0, 1, 4, 9, 16, 25]

# Create a trace
trace = go.Scatter(x=x, y=y, mode='lines+markers', marker=dict(color='blue', size=10))

# Create a layout
layout = go.Layout(title='Line Plot', xaxis=dict(title='X-axis'), yaxis=dict(title='Y-axis'))

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Scatter Plot
Scatter plots are ideal for visualizing relationships between two variables.



```python
import plotly.graph_objs as go

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 11, 12, 13, 14]

# Create a trace
trace = go.Scatter(x=x, y=y, mode='markers', marker=dict(size=15, color='orange'))

# Create a layout
layout = go.Layout(title='Scatter Plot', xaxis=dict(title='X-axis'), yaxis=dict(title='Y-axis'))

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Bar Chart
Bar charts are useful for comparing quantities across categories.

```python
import plotly.graph_objs as go

# Sample data
categories = ['Category A', 'Category B', 'Category C', 'Category D']
values = [23, 45, 56, 78]

# Create a trace
trace = go.Bar(x=categories, y=values, marker=dict(color='purple'))

# Create a layout
layout = go.Layout(title='Bar Chart', xaxis=dict(title='Categories'), yaxis=dict(title='Values'))

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

### Advanced Plot Types
Plotly also supports more advanced plot types, which are useful for specific kinds of data analysis.

#### Histograms
Histograms show the distribution of a dataset.

```python
import plotly.graph_objs as go

# Sample data
data = [1, 2, 2, 3, 3, 3, 4, 4, 5]

# Create a trace
trace = go.Histogram(x=data)

# Create a layout
layout = go.Layout(title='Histogram', xaxis=dict(title='Values'), yaxis=dict(title='Frequency'))

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Box Plots
Box plots are useful for visualizing the spread of data and detecting outliers.

```python
import plotly.graph_objs as go

# Sample data
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 10]

# Create a trace
trace = go.Box(y=data, boxpoints='all', jitter=0.3, pointpos=-1.8)

# Create a layout
layout = go.Layout(title='Box Plot', yaxis=dict(title='Values'))

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Heatmaps
Heatmaps are useful for visualizing matrix-like data or correlations between variables.

```python
import plotly.graph_objs as go

# Sample data
z = [[1, 20, 30],
     [20, 1, 60],
     [30, 60, 1]]

# Create a trace
trace = go.Heatmap(z=z)

# Create a layout
layout = go.Layout(title='Heatmap')

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Surface Plots
Surface plots are 3D plots useful for visualizing three-dimensional data.

```python
import plotly.graph_objs as go

# Sample data
z = [[1, 2, 3], [2, 3, 4], [3, 4, 5]]

# Create a trace
trace = go.Surface(z=z)

# Create a layout
layout = go.Layout(title='3D Surface Plot')

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Subplots and Multiple Plots
Plotly allows you to create multiple plots in a single figure, which is useful for comparing datasets.

```python
from plotly.subplots import make_subplots
import plotly.graph_objs as go

# Create subplots
fig = make_subplots(rows=1, cols=2)

# Add first plot
fig.add_trace(go.Scatter(x=[1, 2, 3], y=[10, 20, 30], mode='lines', name='Line Plot'), row=1, col=1)

# Add second plot
fig.add_trace(go.Bar(x=['A', 'B', 'C'], y=[1, 3, 2], name='Bar Chart'), row=1, col=2)

# Update layout
fig.update_layout(title='Multiple Plots')

# Display the plot
fig.show()
```

### Customizing Visualizations
Customizing your visualizations can greatly improve readability and aesthetic appeal.

#### Layouts
The layout of a plot controls its overall appearance.

```python
import plotly.graph_objs as go

# Sample data
x = [0, 1, 2, 3, 4, 5]
y = [0, 1, 4, 9, 16, 25]

# Create a trace
trace = go.Scatter(x=x, y=y, mode='lines+markers')

# Customize layout
layout = go.Layout(
    title='Custom Layout',
    xaxis=dict(title='X-axis', gridcolor='lightgrey'),
    yaxis=dict(title='Y-axis', gridcolor='lightgrey'),
    plot_bgcolor='rgb(230, 230, 250)',
    paper_bgcolor='rgb(255, 255, 255)',
    font=dict(family='Arial', size=14, color='darkblue')
)

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Annotations
Annotations add text to your plot to highlight specific points.

```python
import plotly.graph_objs as go

# Sample data
x = [0, 1, 2, 3, 4, 5]
y = [0, 1, 4, 9, 16, 25]

# Create a trace
trace = go.Scatter(x=x, y=y, mode='lines+markers')

# Create annotations
annotations = [
    dict(
        x=4, y=16, xref='x', yref='y',
        text='Peak',
        showarrow=True,
        arrowhead=7,
        ax=0, ay=-40
    )
]

# Customize layout
layout = go.Layout(title='Annotated Plot', annotations=annotations)

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Colors and Themes
Plotly provides various color options and themes to style your plots.

```python
import plotly.graph_objs as go

# Sample data
x = ['A', 'B', 'C']
y = [10, 20, 30]

# Create a trace
trace = go.Bar(x=x, y=y, marker=dict(color=['red', 'green', 'blue']))

# Apply a template
layout = go.Layout(title='Custom Colors', template='plotly_dark')

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

### Interactivity
Interactivity is one of the key strengths of Plotly. You can create dynamic plots with hover text, dropdowns, sliders, and animations.

#### Hover Text

```python
import plotly.graph_objs as go

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 17, 18]

# Create a trace with hover text
trace = go.Scatter(
    x=x, y=y, mode='markers+lines',
    marker=dict(color='blue', size=10),
    text=['Point A', 'Point B', 'Point C', 'Point D', 'Point E'],
    hoverinfo='text+x+y'
)

# Create a layout
layout = go.Layout(title='Interactive Hover Text')

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

#### Dropdowns and Sliders
You can add dropdowns and sliders to control different aspects of the plot.

```python
import plotly.graph_objs as go
from plotly.subplots import make_subplots

# Sample data
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 5, 7, 9]
y2 = [2, 4, 6, 8, 10]

# Create traces
trace1 = go.Scatter(x=x, y=y1, mode='lines', name='Dataset 1')
trace2 = go.Scatter(x=x, y=y2, mode='lines', name='Dataset 2')

# Create subplots
fig = make_subplots()

# Add traces
fig.add_trace(trace1)
fig.add_trace(trace2)

# Create dropdown menu
fig.update_layout(
    updatemenus=[
        dict(
            buttons=[
                dict(args=[{'y': [y1]}], label='Dataset 1', method='update'),
                dict(args=[{'y': [y2]}], label='Dataset 2', method='update'),
            ],
            direction='down',
        )
    ]
)

# Create sliders
fig.update_layout(
    sliders=[
        dict(
            steps=[
                dict(method='update', label='1', args=[{'y': [y1]}]),
                dict(method='update', label='2', args=[{'y': [y2]}])
            ],
            active=0
        )
    ]
)

# Display the plot
fig.show()
```

#### Animation
Animations can be used to show changes over time.

```python
import plotly.graph_objs as go
import numpy as np

# Sample data
x = np.linspace(0, 2*np.pi, 100)
frames = [go.Frame(data=[go.Scatter(x=x, y=np.sin(x + i))]) for i in np.linspace(0, 2*np.pi, 20)]

# Create layout
layout = go.Layout(
    title='Animated Sine Wave',
    xaxis=dict(title='X-axis'),
    yaxis=dict(title='Y-axis'),
    updatemenus=[dict(type='buttons', showactive=False, buttons=[dict(label='Play', method='animate', args=[None, dict(frame=dict(duration=100, redraw=True), fromcurrent=True)])])]
)

# Create a figure
fig = go.Figure(
    data=[go.Scatter(x=x, y=np.sin(x))],
    layout=layout,
    frames=frames
)

# Display the plot
fig.show()
```

### 3D Plots
Plotly allows for the creation of 3D plots, which are useful for visualizing data with three dimensions.

```python
import plotly.graph_objs as go
import numpy as np

# Sample data
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

# Create a trace
trace = go.Surface(z=z, x=x, y=y)

# Create a layout
layout = go.Layout(title='3D Surface Plot')

# Create a figure and plot it
fig = go.Figure(data=[trace], layout=layout)
fig.show()
```

### Saving and Exporting Plots
You can save Plotly visualizations as HTML files, images, or JSON.

```python
# Save as HTML
fig.write_html('plot.html')

# Save as a PNG image
fig.write_image('plot.png')

# Save as a JSON file
fig.write_json('plot.json')
```
