# File Handling

## Reading from and writing to files
File handling is a fundamental aspect of programming that allows you to read data from and write data to files on the disk. In Python, file handling is straightforward and versatile, providing multiple ways to interact with files. In this tutorial, we'll explore how to read data from and write data to files in Python, covering file opening modes, reading methods, writing methods, and best practices.

### Opening Files:
Before reading from or writing to a file, you need to open it using the open() function. The open() function takes two arguments: the file path and the mode in which you want to open the file.


```python
file = open("example.txt", "r")  # Open file for reading
```

#### File Opening Modes:
-   "r": Read mode (default). Opens the file for reading.
-   "w": Write mode. Opens the file for writing. If the file does not exist, it creates a new file. If the file exists, it truncates the file.
-   "a": Append mode. Opens the file for writing. If the file does not exist, it creates a new file. If the file exists, it appends data to the end of the file.
-   "r+": Read and write mode. Opens the file for both reading and writing.
-   "b": Binary mode. Opens the file in binary mode (e.g., "rb", "wb").

### Reading from Files:
There are multiple methods available for reading data from files:
-   read(): Reads the entire contents of the file as a single string.
-   readline(): Reads a single line from the file.
-   readlines(): Reads all lines from the file and returns a list of lines.


```python
# Read entire contents of the file
content = file.read()
```

### Writing to Files:
To write data to a file, you use the write() method. Make sure to open the file in write mode ("w" or "a") before writing to it.


```python
with open("example.txt", "w") as file:
    file.write("Hello, World!\n")
    file.write("This is a sample text file.")
```

### Closing Files:
After you're done working with a file, it's essential to close it using the close() method to release system resources associated with the file.


```python
file.close()
```

### Using Context Managers (with Statement):
To ensure that files are properly closed after use, it's recommended to use a context manager (with statement), which automatically closes the file when exiting the block.


```python
with open("example.txt", "r") as file:
    content = file.read()
```

### Best Practices:
-   Use context managers (with statement) to ensure proper file handling and automatic closing of files.
-   Explicitly specify file opening modes ("r", "w", "a") to avoid unintended behavior.
-   Close files explicitly using the close() method after use to release system resources.


## File modes (read, write, append)
File modes specify how files should be opened and manipulated in Python. Understanding file modes is crucial for performing various operations such as reading data from files, writing data to files, and appending data to existing files. In this tutorial, we'll explore the different file modes available in Python—read mode, write mode, and append mode—and how to use them effectively.

### Read Mode ("r"):
Read mode is used to open a file for reading. It allows you to read data from the file but does not allow writing or modifying the file's contents. If the file does not exist, attempting to open it in read mode will raise a FileNotFoundError.

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

### Write Mode ("w"):
Write mode is used to open a file for writing. It creates a new file if the file does not exist, and if the file already exists, it truncates the file (removes its contents) before writing. Be cautious when using write mode, as it can overwrite existing data.


```python
with open("example.txt", "w") as file:
    file.write("Hello, World!\n")
    file.write("This is a sample text file.")
```

### Append Mode ("a"):
Append mode is used to open a file for appending data. It creates a new file if the file does not exist. If the file exists, it appends data to the end of the file without truncating existing content.


```python
with open("example.txt", "a") as file:
    file.write("\nAppending additional text.")
```

### Combining Modes:
You can combine modes to perform both reading and writing operations simultaneously. For example, "r+" opens the file for both reading and writing.


```python
with open("example.txt", "r+") as file:
    content = file.read()
    file.write("\nAppending additional text.")
```

### Best Practices:
-   Use read mode ("r") to open files for reading, write mode ("w") to create or truncate files for writing, and append mode ("a") to append data to existing files.
-   Handle errors gracefully when working with file modes, especially when opening files in read mode.


## Error handling (try-except blocks)
Error handling is a critical aspect of file handling in Python, especially when working with file I/O operations that may encounter various exceptions. By using try-except blocks, you can gracefully handle potential errors that may arise during file operations, such as file not found, permission errors, or unexpected file formats. In this tutorial, we'll explore how to implement error handling using try-except blocks when working with files in Python.
Handling File Not Found Errors:
One of the most common errors encountered when working with files is the FileNotFoundError, which occurs when attempting to open or access a file that does not exist. You can handle this error using a try-except block.

It's essential to handle errors when working with file modes, especially when opening files in read mode to avoid FileNotFoundError exceptions.


```python
try:
    with open("example.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("File not found.")
```

### Handling Permission Errors:
Permission errors may occur when attempting to open a file for reading or writing without the necessary permissions. You can catch PermissionError exceptions to handle such scenarios.


```python
try:
    with open("example.txt", "w") as file:
        file.write("Hello, World!")
except PermissionError:
    print("Permission denied. Unable to write to file.")
```

### Handling Other Exceptions:
In addition to specific file-related exceptions, other exceptions may occur during file operations. You can use a catch-all except block to handle any unexpected exceptions.


```python
try:
    with open("example.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("File not found.")
except Exception as e:
    print("An unexpected error occurred:", e)
```

### Handling Multiple Exceptions:
You can handle multiple exceptions within the same try-except block by specifying multiple except clauses.


```python
try:
    print('success')
except FileNotFoundError:
    print("File not found.")
except PermissionError:
    print("Permission denied.")
```

### Handling Errors in Nested Blocks:
If file operations are nested within other blocks of code, you can still use try-except blocks to handle errors specific to file operations.


```python
try:
    # Other code
    with open("example.txt", "r") as file:
        content = file.read()
        print(content)
    # Other code
except FileNotFoundError:
    print("File not found.")
```

### Best Practices:
-   Use try-except blocks to gracefully handle potential errors that may arise during file operations.
-   Catch specific exceptions (e.g., FileNotFoundError, PermissionError) to provide targeted error messages and handling.
-   Consider using a catch-all except block for unexpected errors to ensure robust error handling.
