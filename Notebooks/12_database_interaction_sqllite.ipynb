## Setting Up SQLite
SQLite comes bundled with Python, so you don't need to install it separately. We'll use the sqlite3 module, which is included in the Python standard library.

## Connecting to a Database
To interact with an SQLite database, you need to connect to it. If the database file doesn't exist, SQLite will create it.


```python
import sqlite3

# Connect to an SQLite database (or create it if it doesn't exist)
connection = sqlite3.connect('example.db')

# Create a cursor object to interact with the database
cursor = connection.cursor()
```

## CRUD operations (Create, Read, Update, Delete)
CRUD stands for Create, Read, Update, and Delete. These are the four basic operations for managing data in a database. This tutorial will guide you through performing CRUD operations with SQLite using Python.
### Creating a Database and Table
To interact with an SQLite database, you need to connect to it. If the database file doesn't exist, SQLite will create it.


```python
import sqlite3

# Connect to an SQLite database (or create it if it doesn't exist)
connection = sqlite3.connect('example.db')

# Create a cursor object to interact with the database
cursor = connection.cursor()

# SQL command to create a table
create_table_query = '''
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT
)
'''

# Execute the SQL command
cursor.execute(create_table_query)

# Commit the changes
connection.commit()
```

### Inserting Data (Create)
You can insert data into a table using the INSERT INTO SQL command.


```python
# SQL command to insert data
insert_data_query = '''
INSERT INTO employees (name, age, department)
VALUES (?, ?, ?)
'''

# Data to be inserted
employees = [
    ('Alice', 30, 'HR'),
    ('Bob', 25, 'Engineering'),
    ('Charlie', 35, 'Marketing')
]

# Execute the SQL command for each entry
cursor.executemany(insert_data_query, employees)

# Commit the changes
connection.commit()
```

### Querying Data (Read)
You can retrieve data from the database using the SELECT SQL command.


```python
# SQL command to query data
select_query = 'SELECT * FROM employees'

# Execute the SQL command
cursor.execute(select_query)

# Fetch all results
rows = cursor.fetchall()

# Print the results
for row in rows:
    print(row)
```

### Updating Data (Update)
You can update existing data using the UPDATE SQL command.


```python
# SQL command to update data
update_query = '''
UPDATE employees
SET age = ?
WHERE name = ?
'''

# Data to be updated
updated_data = (28, 'Bob')

# Execute the SQL command
cursor.execute(update_query, updated_data)

# Commit the changes
connection.commit()

# Verify the update
cursor.execute(select_query)
rows = cursor.fetchall()
for row in rows:
    print(row)
```

### Deleting Data (Delete)
You can delete data from the database using the DELETE SQL command.


```python
# SQL command to delete data
delete_query = '''
DELETE FROM employees
WHERE name = ?
'''

# Data to specify which record to delete
name_to_delete = ('Charlie',)

# Execute the SQL command
cursor.execute(delete_query, name_to_delete)

# Commit the changes
connection.commit()

# Verify the deletion
cursor.execute(select_query)
rows = cursor.fetchall()
for row in rows:
    print(row)
```

## Querying data with SQL
Querying data involves retrieving specific information from a database using SQL (Structured Query Language). This tutorial will cover various types of SQL queries and how to execute them using Python.
### Basic Select Queries
A SELECT query retrieves data from a database. The simplest form retrieves all rows and columns from a table.


```python
# SQL command to query data
select_query = 'SELECT * FROM employees'

# Execute the SQL command
cursor.execute(select_query)

# Fetch all results
rows = cursor.fetchall()

# Print the results
print("All employees:")
for row in rows:
    print(row)
```

### Filtering Results
You can filter results using the WHERE clause.


```python
# SQL command to query data with a condition
select_query = 'SELECT * FROM employees WHERE department = ?'
department = ('Engineering',)

# Execute the SQL command
cursor.execute(select_query, department)

# Fetch all results
rows = cursor.fetchall()

# Print the results
print("\nEmployees in Engineering:")
for row in rows:
    print(row)
```

### Sorting Results
You can sort results using the ORDER BY clause.


```python
# SQL command to query and sort data
select_query = 'SELECT * FROM employees ORDER BY age DESC'

# Execute the SQL command
cursor.execute(select_query)

# Fetch all results
rows = cursor.fetchall()

# Print the results
print("\nEmployees sorted by age (descending):")
for row in rows:
    print(row)
```

### Limiting Results
You can limit the number of results using the LIMIT clause.


```python
# SQL command to query and limit data
select_query = 'SELECT * FROM employees ORDER BY age DESC LIMIT 2'

# Execute the SQL command
cursor.execute(select_query)

# Fetch all results
rows = cursor.fetchall()

# Print the results
print("\nTop 2 oldest employees:")
for row in rows:
    print(row)
```

### Using Aggregate Functions
Aggregate functions perform a calculation on a set of values and return a single value. Common aggregate functions include COUNT, SUM, AVG, MIN, and MAX.


```python
# SQL command to count the number of employees
count_query = 'SELECT COUNT(*) FROM employees'

# Execute the SQL command
cursor.execute(count_query)

# Fetch the result
count = cursor.fetchone()[0]
print(f"\nTotal number of employees: {count}")

# SQL command to calculate the average age of employees
avg_age_query = 'SELECT AVG(age) FROM employees'

# Execute the SQL command
cursor.execute(avg_age_query)

# Fetch the result
avg_age = cursor.fetchone()[0]
print(f"Average age of employees: {avg_age:.2f}")

# SQL command to find the youngest employee
min_age_query = 'SELECT MIN(age) FROM employees'

# Execute the SQL command
cursor.execute(min_age_query)

# Fetch the result
min_age = cursor.fetchone()[0]
print(f"Youngest employee age: {min_age}")

# SQL command to find the oldest employee
max_age_query = 'SELECT MAX(age) FROM employees'

# Execute the SQL command
cursor.execute(max_age_query)

# Fetch the result
max_age = cursor.fetchone()[0]
print(f"Oldest employee age: {max_age}")
```

### Query Using a Window Function
Now, let's use the ROW_NUMBER() function to assign a unique number to each row within a partition of the result set.


```python
# SQL command to use the ROW_NUMBER() window function
select_query = '''
SELECT
    name,
    department,
    age,
    ROW_NUMBER() OVER (PARTITION BY department ORDER BY age DESC) AS row_num
FROM employees
'''

# Execute the SQL command
cursor.execute(select_query)

# Fetch all results
rows = cursor.fetchall()

# Print the results
print("Employees with row numbers within their departments:")
for row in rows:
    print(row)

# Close the connection
connection.close()
```

#### Using RANK() and DENSE_RANK()
The RANK() and DENSE_RANK() functions can be used similarly to rank rows within partitions.


```python
# SQL command to use the RANK() window function
select_query = '''
SELECT
    name,
    department,
    age,
    RANK() OVER (PARTITION BY department ORDER BY age DESC) AS rank
FROM employees
'''

# Execute the SQL command
cursor.execute(select_query)

# Fetch all results
rows = cursor.fetchall()

# Print the results
print("Employees with ranks within their departments:")
for row in rows:
    print(row)
```

#### Using LAG() and LEAD()
The LAG() and LEAD() functions allow you to access data from previous or next rows in the result set.


```python
# SQL command to use the LAG() window function
select_query = '''
SELECT
    name,
    department,
    age,
    LAG(age, 1) OVER (PARTITION BY department ORDER BY age) AS previous_age
FROM employees
'''

# Execute the SQL command
cursor.execute(select_query)

# Fetch all results
rows = cursor.fetchall()

# Print the results
print("Employees with their previous age in the same department:")
for row in rows:
    print(row)
```
