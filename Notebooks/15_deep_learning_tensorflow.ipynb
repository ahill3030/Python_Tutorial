## Introduction to neural networks
### What are Neural Networks?
Neural networks are a class of machine learning models inspired by the structure and function of the human brain. They consist of interconnected layers of artificial neurons (also called nodes or units) that process and transform input data to produce output predictions or decisions. Neural networks are capable of learning complex patterns and relationships from data through a process known as training.

### Biological Inspiration
The design of neural networks is inspired by the structure and function of biological neurons in the human brain. In a biological neuron, signals (or inputs) from other neurons are received through dendrites, processed in the cell body, and then transmitted as output signals (or activations) through the axon to other neurons. Similarly, artificial neurons in a neural network receive input signals, apply transformations using learned parameters, and produce output activations.

### Basic Components of a Neural Network
A neural network consists of several basic components, including:
-   Input Layer: The input layer receives input data and passes it to the next layer.
-   Hidden Layers: Hidden layers are intermediate layers between the input and output layers. Each hidden layer consists of multiple neurons that perform transformations on the input data.
-   Output Layer: The output layer produces the final predictions or decisions based on the output activations of the last hidden layer.
-   Connections (Edges): Connections between neurons represent weighted connections, which determine how input signals are combined and transformed by each neuron.
-   Weights and Biases: Weights represent the strength of connections between neurons, while biases represent the offset or baseline activation of each neuron.
### Activation Functions
Activation functions introduce non-linearity into the neural network, enabling it to learn complex patterns and relationships in the data. Common activation functions include:
-   Sigmoid: S-shaped curve that squashes input values between 0 and 1.
-   ReLU (Rectified Linear Unit): Linear for positive input values and zero for negative input values.
-   Tanh (Hyperbolic Tangent): S-shaped curve similar to sigmoid but squashes input values between -1 and 1.
### Types of Neural Networks
There are various types of neural networks, each designed for specific tasks and applications. Some common types of neural networks include:
-   Feedforward Neural Networks (FNN): Basic type of neural network where information flows in one direction from input to output.
-   Convolutional Neural Networks (CNN): Designed for processing structured grid-like data, such as images.
-   Recurrent Neural Networks (RNN): Designed for processing sequential data, such as time series or text data.
-   Generative Adversarial Networks (GAN): Comprising two neural networks (generator and discriminator) that compete against each other to generate realistic data samples.

## Image classification example
Image classification is a common task in computer vision where the goal is to classify images into predefined categories or classes. In this tutorial, we'll provide an example of how to perform image classification using deep learning with TensorFlow and Keras.

### Overview of Image Classification
Image classification is the process of assigning a label or category to an image based on its visual content. It is a fundamental task in computer vision and has numerous applications, including object detection, facial recognition, and medical image analysis.

### Dataset Preparation
For this example, we'll use the CIFAR-10 dataset, which consists of 60,000 32x32 color images in 10 classes, with 6,000 images per class. We'll download and preprocess the dataset using TensorFlow's built-in datasets module.


```python
import tensorflow as tf

# Load the CIFAR-10 dataset
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.cifar10.load_data()

# Normalize pixel values to between 0 and 1
train_images, test_images = train_images / 255.0, test_images / 255.0
```

### Building the Deep Learning Model
We'll build a convolutional neural network (CNN) model for image classification. CNNs are well-suited for image data and are commonly used in computer vision tasks.


```python
from tensorflow.keras import layers, models

# Define the CNN model
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10)
])
```

### Compiling the Model
Next, we'll compile the model by specifying the loss function, optimizer, and evaluation metric.


```python
# Compile the model
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])
```

### Training the Model
We'll train the model on the training data for a fixed number of epochs.


```python
# Train the model
history = model.fit(train_images, train_labels, epochs=10, validation_data=(test_images, test_labels))
```

### Evaluating the Model
After training the model, we'll evaluate its performance on the test data.


```python
# Evaluate the model
test_loss, test_acc = model.evaluate(test_images, test_labels)
print("Test Accuracy:", test_acc)
```

## Text generation example
Text generation is a fascinating application of deep learning where the goal is to generate new text based on a given input or context. In this tutorial, we'll provide an example of how to perform text generation using deep learning with TensorFlow and Keras.

### Overview of Text Generation
Text generation involves training a model to predict the next word or sequence of words in a text based on the previous context. Recurrent Neural Networks (RNNs) and Long Short-Term Memory (LSTM) networks are commonly used for text generation tasks due to their ability to capture sequential patterns in data.

### Dataset Preparation
For this example, we'll use a collection of text data, such as a corpus of novels, articles, or poems. You can use any text data source you prefer. We'll preprocess the text data to convert it into sequences of fixed length that the model can learn from.


```python
import tensorflow as tf

# Load the text data
text = open('text_data.txt', 'r').read()

# Tokenize the text into words
tokenizer = tf.keras.preprocessing.text.Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1

# Convert text into sequences of fixed length
input_sequences = []
for line in text.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)
```

### Building the Deep Learning Model
We'll build a simple LSTM-based model for text generation.


```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Pad sequences to ensure uniform length
max_sequence_len = max([len(seq) for seq in input_sequences])
input_sequences = tf.keras.preprocessing.sequence.pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')

# Create predictors and labels
predictors, label = input_sequences[:, :-1], input_sequences[:, -1]

# Define the LSTM model
model = Sequential([
    Embedding(total_words, 100, input_length=max_sequence_len-1),
    LSTM(150),
    Dense(total_words, activation='softmax')
])
```

### Compiling the Model
Next, we'll compile the model by specifying the loss function and optimizer.


```python
# Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

### Training the Model
We'll train the model on the prepared sequences of text data.


```python
# Train the model
history = model.fit(predictors, label, epochs=100, verbose=1)
```

### Generating Text
After training the model, we can generate new text by providing a seed text as input and predicting the next word iteratively.


```python
# Function to generate text
def generate_text(seed_text, next_words, model, max_sequence_len):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = tf.keras.preprocessing.sequence.pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')
        predicted = model.predict_classes(token_list, verbose=0)
        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break
        seed_text += " " + output_word
    return seed_text

# Generate text
generated_text = generate_text("I", 10, model, max_sequence_len)
print(generated_text)
```

## Neural networks using Torch:
PyTorch is a popular deep learning library in Python known for its flexibility and dynamic computation graph. In this tutorial, we'll provide an example of how to build and train a neural network using PyTorch for a classification task.
### Overview of PyTorch
PyTorch is an open-source machine learning library developed by Facebook's AI research lab. It provides a flexible and dynamic framework for building and training deep learning models. PyTorch's dynamic computation graph allows for more intuitive model building and debugging compared to static graph frameworks like TensorFlow.

### Dataset Preparation
For this example, we'll use the CIFAR-10 dataset, which consists of 60,000 32x32 color images in 10 classes. We'll use PyTorch's built-in dataset module to load and preprocess the dataset.


```python
import torch
import torchvision
import torchvision.transforms as transforms

# Define data transformations
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

# Load the CIFAR-10 dataset
trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=4, shuffle=True, num_workers=2)

testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=4, shuffle=False, num_workers=2)

classes = ('plane', 'car', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck')
```

### Building the Neural Network Model
We'll build a simple convolutional neural network (CNN) model for image classification using PyTorch's nn module.


```python
import torch.nn as nn
import torch.nn.functional as F

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.conv1 = nn.Conv2d(3, 6, 5)
        self.pool = nn.MaxPool2d(2, 2)
        self.conv2 = nn.Conv2d(6, 16, 5)
        self.fc1 = nn.Linear(16 * 5 * 5, 120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(-1, 16 * 5 * 5)
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x

net = Net()
```

### Training the Model
We'll define a loss function and optimizer, and then train the model on the training data.


```python
import torch.optim as optim

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.parameters(), lr=0.001, momentum=0.9)

for epoch in range(2):
    running_loss = 0.0
    for i, data in enumerate(trainloader, 0):
        inputs, labels = data
        optimizer.zero_grad()
        outputs = net(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
        if i % 2000 == 1999:
            print('[%d, %5d] loss: %.3f' %
                  (epoch + 1, i + 1, running_loss / 2000))
            running_loss = 0.0

print('Finished Training')
```

### Evaluating the Model
After training the model, we'll evaluate its performance on the test data.


```python
correct = 0
total = 0
with torch.no_grad():
    for data in testloader:
        images, labels = data
        outputs = net(images)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print('Accuracy of the network on the 10000 test images: %d %%' % (
    100 * correct / total))
```
