## Variables and data types
Variables are used to store data values in memory.
In Python, variables are created when you assign a value to them using the assignment operator =.
Variable names can contain letters, numbers, and underscores but cannot start with a number.
Python is dynamically typed, meaning you don't need to declare the data type of a variable explicitly.



```python
# Variable assignment
x = 10
name = "John"

# Printing variables
print(x)    # Output: 10
print(name) # Output: John
```

### Data Types:
Python supports various data types, including:
-   Numeric Types:
    -   int: Integer numbers (e.g., 10, -5, 1000)
    -   float: Floating-point numbers with a decimal point (e.g., 3.14, -0.5, 2.0)
    -   complex: Complex numbers with a real and imaginary part (e.g., 3 + 4j)
- Sequence Types:
    -   str: Strings of characters (e.g., "hello", 'Python', "123")
    -   list: Ordered collection of items (e.g., [1, 2, 3], ['a', 'b', 'c'])
    -   tuple: Immutable ordered collection of items (e.g., (1, 2, 3), ('a', 'b', 'c'))
-   Boolean Type:
    -   bool: Boolean values True or False
-   Set Types:
    -   set: Unordered collection of unique items (e.g., {1, 2, 3}, {'a', 'b', 'c'})
    -   frozenset: Immutable set
-   Mapping Type:
    -   dict: Collection of key-value pairs (e.g., {'name': 'John', 'age': 30})

#### Type Conversion (Casting):
Python allows you to convert between different data types using type conversion functions.
int(), float(), str(), list(), tuple(), set(), dict(), etc., are used for type conversion.


```python
# Type conversion
x = 10
y = float(x)  # Convert integer to float
z = str(x)    # Convert integer to string
```

#### Checking Data Type:
You can check the data type of a variable using the type() function.

```python
x = 10
print(type(x))  # Output: <class 'int'>
```

### Python Interpreter and Interactive Mode:
Python can be executed in two modes: interactive mode and script mode.
Interactive mode allows you to enter Python commands directly into the interpreter and see the results immediately.
To start the Python interpreter in interactive mode, simply open your terminal or command prompt and type python. You'll see the Python prompt (>>>) where you can enter commands.


```python
$ python
Python 3.9.7 (default, Sep  3 2021, 21:19:12)
[GCC 7.5.0] on linux
>>> print("Hello, Python!")
Hello, Python!
```

### Python Comments:
Comments are used to add explanations or notes within the code.
In Python, comments start with the # symbol and continue until the end of the line.
Comments are ignored by the Python interpreter and are not executed.
Single-line comments:
•	Single-line comments start with the # symbol and continue until the end of the line. They are typically used for brief explanations or notes within the code.


```python
# This is a single-line comment
print("Hello, Python!")  # This is another single-line comment
```

#### Multi-Line Comments (Using Triple Quotes):
Python doesn't have built-in multi-line comment syntax like some other programming languages. However, you can achieve multi-line comments by using triple quotes (''' or """). This is often used for longer explanations or comments.


```python
'''
This is a multi-line comment.
It spans across multiple lines.
'''

"""
Another way to create a multi-line comment.
You can use either single or double quotes.
"""
```

## Basic input/output
Displaying Output with print() Function:
The print() function is used to display output on the console. It takes one or more arguments and prints them to the standard output, usually the console window.


```python
print("Hello, Python!")  # Output: Hello, Python!
```

### Reading User Input with input() Function:
The input() function is used to read input from the user through the console. It prompts the user for input and waits for them to enter data followed by the Enter key.
The input() function always returns a string, so you may need to convert it to the appropriate data type if necessary.


```python
name = input("Enter your name: ")
print("Hello,", name)
```

### Formatted Output with f-strings:
Python 3.6 introduced f-strings (formatted string literals), which provide a concise and readable way to format strings with variables or expressions.
In f-strings, you can embed Python expressions within curly braces {}.


```python
name = "John"
age = 30
print(f"Hello, {name}! You are {age} years old.")
```

### Working with Files:
Python provides built-in functions for reading from and writing to files. You can open files in different modes such as read mode ('r'), write mode ('w'), append mode ('a'), etc.
The with statement is used to automatically close the file when done, ensuring proper resource management.
Remember to handle file exceptions (e.g., FileNotFoundError) when working with files.


```python
# Writing to a file
with open("output.txt", "w") as file:
    file.write("Hello, Python!\n")

# Reading from a file
with open("output.txt", "r") as file:
    content = file.read()
    print(content)  # Output: Hello, Python!
```

### Operators and expressions
Operators are symbols that perform operations on variables and values in Python. They are used to manipulate data, perform calculations, compare values, and more. In this tutorial section, we'll cover the various types of operators in Python, including arithmetic, assignment, comparison, logical, and bitwise operators.

#### Arithmetic Operators:
Arithmetic operators are used to perform mathematical operations such as addition, subtraction, multiplication, division, etc.


```python
x = 10
y = 5

addition = x + y        # Addition
subtraction = x - y     # Subtraction
multiplication = x * y  # Multiplication
division = x / y        # Division
modulus = x % y         # Modulus (remainder)
exponentiation = x ** y # Exponentiation
```

#### Assignment Operators:
Assignment operators are used to assign values to variables. They combine the assignment (=) operator with other arithmetic operators.


```python
x = 10
x += 5  # Equivalent to x = x + 5
x -= 3  # Equivalent to x = x - 3
x *= 2  # Equivalent to x = x * 2
x /= 4  # Equivalent to x = x / 4
```

#### Comparison Operators:
Comparison operators are used to compare the values of two variables or expressions. They return a Boolean value (True or False).


```python
x = 10
y = 5

greater_than = x > y       # Greater than
less_than = x < y          # Less than
equal_to = x == y          # Equal to
not_equal_to = x != y      # Not equal to
greater_than_or_equal_to = x >= y  # Greater than or equal to
less_than_or_equal_to = x <= y     # Less than or equal to
```

#### Logical Operators:
Logical operators are used to combine conditional statements. They include and, or, and not.


```python
x = True
y = False

logical_and = x and y  # Logical AND
logical_or = x or y    # Logical OR
logical_not = not x    # Logical NOT
```

#### Bitwise Operators:
Bitwise operators perform bitwise operations on binary representations of integers.


```python
x = 10   # Binary: 1010
y = 5    # Binary: 0101

bitwise_and = x & y   # Bitwise AND (1010 & 0101 = 0000)
bitwise_or = x | y    # Bitwise OR (1010 | 0101 = 1111)
bitwise_xor = x ^ y   # Bitwise XOR (1010 ^ 0101 = 1111)
bitwise_not_x = ~x    # Bitwise NOT (Inverts the bits)
bitwise_left_shift = x << 1  # Bitwise left shift by 1 (1010 << 1 = 10100)
bitwise_right_shift = x >> 1 # Bitwise right shift by 1 (1010 >> 1 = 0101)
```

## Control flow (if statements, loops)
Control flow structures allow you to control the flow of execution in your Python programs. They enable you to make decisions based on conditions and repeat actions multiple times. In this tutorial section, we'll cover if statements for conditional execution and various types of loops for repetitive tasks in Python.

### If Statements:
-   If statements are used to execute a block of code only if a specified condition is true.
-   If the condition is true, the code block under the if statement is executed.


```python
x = 10

if x > 0:
    print("x is positive")
```

Optionally, you can use elif (short for "else if") and else clauses to specify alternative conditions.

```python
if x > 0:
    print("x is positive")
elif x == 0:
    print("x is zero")
else:
    print("x is negative")
```

### Loops:

#### For Loop:
The for loop is used to iterate over a sequence (such as a list, tuple, string, or range) and execute a block of code for each element.


```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

The fruit variable takes on each value in the fruits list in sequence.
You can also use the range() function to generate a sequence of numbers to iterate over.


```python
for i in range(5):
    print(i)
```

#### While Loop:
The while loop is used to execute a block of code repeatedly as long as a specified condition is true.


```python
i = 0

while i < 5:
    print(i)
    i += 1
```

The loop continues as long as the condition i < 5 is true.
Be cautious with while loops to avoid infinite loops.


### Loop Control Statements:
Python provides loop control statements to change the execution flow within loops.

#### break: 
Terminates the loop prematurely when a certain condition is met.


```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

#### continue: 
Skips the rest of the current iteration and proceeds to the next iteration of the loop.

```python
for i in range(10):
    if i == 5:
        continue
    print(i)
```

#### pass: 
Acts as a placeholder and does nothing. It is often used when a statement is syntactically required but you don't want any code to execute.

```python
x = 10

if x < 0:
    print("Negative")
elif x == 0:
    pass
else:
    print("Positive")
```
