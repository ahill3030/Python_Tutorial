# **Tutorial: Getting Started with Pydantic**

Pydantic is a Python library that helps with **data validation** and **data parsing** using Python type annotations. It’s particularly useful for applications where you need to ensure the data you’re working with adheres to specific types and structures, such as API data processing, form validation, or configuration handling.

This tutorial will guide you through the basics of using Pydantic, from installing the library to creating models, parsing data, and handling advanced use cases.

---

## **1. Installation**

Install Pydantic using pip:

```bash
pip install pydantic
```

---

## **2. Basics of Pydantic**

### **Creating a Model**
Pydantic's core concept is the `BaseModel`. You define your data structure as a Python class that inherits from `BaseModel`.

```python
from pydantic import BaseModel

class User(BaseModel):
    id: int
    name: str
    email: str
    is_active: bool = True  # default value
```

### **Instantiating and Validating a Model**
Pydantic automatically validates data when you create an instance of a model.

```python
data = {
    "id": 1,
    "name": "John Doe",
    "email": "john.doe@example.com"
}

user = User(**data)
print(user)
# Output: id=1 name='John Doe' email='john.doe@example.com' is_active=True
```

If the data is invalid, Pydantic raises a `ValidationError`.

```python
invalid_data = {"id": "one", "name": "Jane"}
try:
    user = User(**invalid_data)
except Exception as e:
    print(e)
```

**Output:**
```
1 validation error for User
id
  value is not a valid integer (type=type_error.integer)
email
  field required (type=value_error.missing)
```

## **3. Data Parsing**

Pydantic can parse different data types into the correct format.

```python
from datetime import datetime

class Event(BaseModel):
    name: str
    date: datetime

event = Event(name="Conference", date="2024-11-27")
print(event)
# Output: name='Conference' date=datetime.datetime(2024, 11, 27, 0, 0)
```

---

## **4. Model Methods**

### `dict()`
Converts the model instance into a dictionary.

```python
print(user.dict())
# Output: {'id': 1, 'name': 'John Doe', 'email': 'john.doe@example.com', 'is_active': True}
```

### `json()`
Serializes the model instance to JSON.

```python
print(user.json())
# Output: {"id": 1, "name": "John Doe", "email": "john.doe@example.com", "is_active": true}
```

### `parse_obj()`
Creates a model from a dictionary.

```python
user = User.parse_obj(data)
```

### `parse_raw()`
Parses data from a JSON string.

```python
json_data = '{"id": 2, "name": "Alice", "email": "alice@example.com"}'
user = User.parse_raw(json_data)
```

## **5. Advanced Features**

### **Validators**
You can add custom validation logic using the `@validator` decorator.

```python
from pydantic import validator

class User(BaseModel):
    id: int
    email: str

    @validator("email")
    def validate_email(cls, v):
        if "@" not in v:
            raise ValueError("Invalid email address")
        return v
```

### **Nested Models**
Models can include other models as fields.

```python
class Address(BaseModel):
    street: str
    city: str

class UserWithAddress(BaseModel):
    id: int
    name: str
    address: Address

data = {
    "id": 1,
    "name": "John Doe",
    "address": {"street": "123 Main St", "city": "Springfield"}
}

user = UserWithAddress(**data)
print(user.address.city)  # Output: Springfield
```

### **Field Customization**
You can use the `Field` function to customize fields.

```python
from pydantic import Field

class Product(BaseModel):
    name: str
    price: float = Field(..., gt=0, description="Price must be positive")

product = Product(name="Laptop", price=1000)
print(product)
```

## **6. Using Pydantic for Configuration**
Pydantic models can be used to manage application configurations.

```python
from pydantic import BaseSettings

class Config(BaseSettings):
    app_name: str
    debug: bool = False
    database_url: str

    class Config:
        env_file = ".env"

config = Config()
print(config.app_name)
```

## **7. Error Handling**

Pydantic provides detailed error messages:

```python
try:
    User(id=1, name="Jane", email="invalid-email")
except ValidationError as e:
    print(e.json())
```

**Output:**
```json
[
  {
    "loc": ["email"],
    "msg": "value is not a valid email address",
    "type": "value_error.email"
  }
]
```

## **8. Tips for Best Practices**

- Use type annotations wherever possible to leverage Pydantic's full power.
- Define default values or use the `Field` function for more control over fields.
- Handle `ValidationError` gracefully in production systems.
