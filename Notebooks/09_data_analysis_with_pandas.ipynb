# Data Analysis with Pandas
Pandas is an open-source data manipulation and analysis library for Python, built on top of NumPy. It provides data structures and functions needed to work with structured data seamlessly and efficiently. This tutorial will cover the basics of using Pandas for data manipulation and analysis.


## Introduction to Pandas DataFrame
Pandas is designed to handle tabular data (similar to SQL tables or Excel spreadsheets) and offers powerful, flexible, and easy-to-use data structures for data manipulation and analysis.

### Installing Pandas


```python
import pandas as pd
import numpy as np
```

### Data Structures in Pandas
#### Series: 
A one-dimensional array-like object containing an array of data and an associated array of data labels (index).


```python
# Creating a Series
s = pd.Series([1, 3, 5, np.nan, 6, 8])
print(s)
```

#### DataFrame: 
A two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

### Reading and writing data
Pandas can read data from various file formats such as CSV, Excel, SQL, and more.


```python
# Reading a CSV file
df = pd.read_csv('data.csv')

# Reading an Excel file
df = pd.read_excel('data.xlsx', sheet_name='Sheet1')
```

## Data Exploration
Exploring the dataset to understand its structure and content.


```python
# Display the first few rows
print(df.head())

# Display the last few rows
print(df.tail())

# Display the shape of the DataFrame (rows, columns)
print(df.shape)

# Display the column names
print(df.columns)

# Display the data types of each column
print(df.dtypes)

# Summary statistics
print(df.describe())

# Check for missing values
print(df.isnull().sum())
```

### Data Cleaning
Cleaning the data by handling missing values, duplicates, and incorrect data types.

#### Handling Missing Values:


```python
# Drop rows with missing values
df.dropna(inplace=True)

# Fill missing values
df.fillna(value=0, inplace=True)
```

#### Removing Duplicates:

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

### Converting Data Types:

```python
# Convert a column to a different data type
df['column_name'] = df['column_name'].astype('int')
```

## Data Manipulation
Manipulating data by selecting, filtering, and transforming the DataFrame.

### Selecting Columns:


```python
# Selecting a single column
print(df['A'])

# Selecting multiple columns
print(df[['A', 'B']])
```

### Adding New Columns:

```python
# Adding a new column
df['D'] = df['A'] + df['B']
print(df)
```

### Renaming Columns:

```python
# Renaming columns
df.rename(columns={'A': 'a', 'B': 'b'}, inplace=True)
print(df)
```

### Grouping and Aggregation
Grouping data and performing aggregate functions on the groups.


```python
# Grouping by a column and calculating the mean
grouped = df.groupby('A').mean()
print(grouped)

# Aggregating with multiple functions
agg_df = df.groupby('A').agg({'B': ['mean', 'sum'], 'C': 'max'})
print(agg_df)
```

## Merging and Joining Data
Combining multiple DataFrames using merge, join, and concatenate operations.

### Merging DataFrames:


```python
# Merging two DataFrames on a common column
df1 = pd.DataFrame({'key': ['A', 'B', 'C', 'D'],
                    'value': [1, 2, 3, 4]})
df2 = pd.DataFrame({'key': ['B', 'D', 'E', 'F'],
                    'value': [5, 6, 7, 8]})

merged_df = pd.merge(df1, df2, on='key', how='inner')
print(merged_df)
```

### Joining DataFrames:

```python
# Joining two DataFrames
df1 = df1.set_index('key')
df2 = df2.set_index('key')

joined_df = df1.join(df2, lsuffix='_left', rsuffix='_right')
print(joined_df)
```

### Concatenating DataFrames:

```python
# Concatenating DataFrames along rows
concat_df = pd.concat([df1, df2], axis=0)
print(concat_df)

# Concatenating DataFrames along columns
concat_df = pd.concat([df1, df2], axis=1)
print(concat_df)
```

## Time Series Analysis
Pandas provides powerful tools to work with time series data.

### Parsing Dates:


```python
# Reading a CSV file and parsing dates
df = pd.read_csv('time_series_data.csv', parse_dates=['date'])
```

### Setting the Index:

```python
# Setting the date column as the index
df.set_index('date', inplace=True)
```

### Resampling Time-Series Data:

```python
# Resampling the time series data to a different frequency
monthly_df = df.resample('M').mean()
print(monthly_df)
```

### Time-Series Plotting:

```python
import matplotlib.pyplot as plt

# Plotting the time series data
df['value'].plot()
plt.show()
```

## Data manipulation (filtering, sorting, grouping)
Pandas is a powerful library for data manipulation and analysis in Python. This tutorial will focus on three key aspects of data manipulation using Pandas: filtering, sorting, and grouping. Understanding these concepts is crucial for effectively managing and analyzing your data.

### Filtering Rows:

```python
# Filtering rows based on a condition
print(df[df['A'] > 2])

# Filtering rows based on multiple conditions
print(df[(df['A'] > 2) & (df['B'] < 8)])
```

### Basic Filtering:

```python
# Filter rows where Age is greater than 25
filtered_df = df[df['Age'] > 25]
print(filtered_df)
```

### Filtering with Multiple Conditions:

```python
# Filter rows where Age is greater than 25 and Salary is less than 62000
filtered_df = df[(df['Age'] > 25) & (df['Salary'] < 62000)]
print(filtered_df)
```

### Filtering using String Methods:

```python
# Filter rows where the Department is 'HR'
filtered_df = df[df['Department'] == 'HR']
print(filtered_df)

# Filter rows where the Name starts with 'A'
filtered_df = df[df['Name'].str.startswith('A')]
print(filtered_df)
```

### Filtering using `query` method:

```python
# Using the query method to filter data
filtered_df = df.query('Age > 25 and Salary < 62000')
print(filtered_df)
```

## Sorting Data
Sorting is used to arrange the data in a particular order, either ascending or descending.

### Sorting by a Single Column:


```python
# Sort by Age in ascending order
sorted_df = df.sort_values(by='Age')
print(sorted_df)

# Sort by Age in descending order
sorted_df = df.sort_values(by='Age', ascending=False)
print(sorted_df)
```

### Sorting by Multiple Columns:

```python
# Sort by Department and then by Salary within each department
sorted_df = df.sort_values(by=['Department', 'Salary'])
print(sorted_df)

# Sort by Department (ascending) and then by Salary (descending)
sorted_df = df.sort_values(by=['Department', 'Salary'], ascending=[True, False])
print(sorted_df)
```

## Grouping Data
Grouping is a powerful method in Pandas for performing aggregate operations based on the unique values in one or more columns.

### Basic Grouping:


```python
# Group by Department and calculate the mean age and salary
grouped_df = df.groupby('Department').mean()
print(grouped_df)
```

### Grouping and Aggregating with Multiple Functions:

```python
# Group by Department and calculate multiple aggregate statistics
grouped_df = df.groupby('Department').agg({
    'Age': ['mean', 'min', 'max'],
    'Salary': ['mean', 'sum']
})
print(grouped_df)
```

### Grouping with Custom Functions:

```python
# Define a custom aggregation function
def range_func(x):
    return x.max() - x.min()

# Group by Department and apply the custom function
grouped_df = df.groupby('Department').agg({
    'Age': range_func,
    'Salary': 'sum'
})
print(grouped_df)
```

### Iterating over groups:

```python
# Define a custom aggregation function
def range_func(x):
    return x.max() - x.min()

# Group by Department and apply the custom function
grouped_df = df.groupby('Department').agg({
    'Age': range_func,
    'Salary': 'sum'
})
print(grouped_df)
```
